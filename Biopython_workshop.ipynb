```python
#creating a list

gene = ["GeneA", "GeneB", "GeneC", "GeneD"]
counts = [10, 25, 7, 15]

#we are using pandas
#creating a dataframe - organises raw data

data = pd.DataFrame({"Gene":gene, "Counts":counts})
print(data)

#barplot

sns.barplot(x="Gene", y="Counts", data = data, palette="viridis")

#add title and labels
plt.title("Gene Expression - ABCD Bar Plot")
plt.xlabel("Gene")
plt.ylabel("Expression Count")

#show plot

plt.show()
```

```python
from Bio.Seq import Seq

# Example DNA sequence
dna_sequence = "ATGGCCATTGTAATGGGCCGCTGAAAGGGTGCCCGATAG"

# Create a Seq object
dna_seq = Seq(dna_sequence)

# Calculate GC content
gc_content = float(dna_seq.count("G") + dna_seq.count("C")) / len(dna_seq) * 100

# Get reverse complement
rev_comp = dna_seq.reverse_complement()

print("DNA Sequence:       ", dna_seq)
print("GC Content (%):     ", round(gc_content, 2))
print("Reverse Complement: ", rev_comp)
```

```python
from Bio.Seq import Seq

dna_sequence = "TGGCCATTGTAATGGGCCGCTGAAAGGGTGCCCGATA"

#creating a Seq object
dna_seq = Seq(dna_sequence)

#translating dna into protein
prot_seq = dna_seq.translate()

print("DNA sequence: ", dna_seq)
print("Protein sequence: ", prot_seq)
```
