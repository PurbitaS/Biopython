```python
from Bio.Seq import Seq

# Example DNA sequence
dna_sequence = "ATGGCCATTGTAATGGGCCGCTGAAAGGGTGCCCGATAG"

# Create a Seq object
dna_seq = Seq(dna_sequence)

# Calculate GC content
gc_content = float(dna_seq.count("G") + dna_seq.count("C")) / len(dna_seq) * 100

# Get reverse complement
rev_comp = dna_seq.reverse_complement()

print("DNA Sequence:       ", dna_seq)
print("GC Content (%):     ", round(gc_content, 2))
print("Reverse Complement: ", rev_comp)
```

```python
from Bio.Seq import Seq

dna_sequence = "TGGCCATTGTAATGGGCCGCTGAAAGGGTGCCCGATA"

#creating a Seq object
dna_seq = Seq(dna_sequence)

#translating dna into protein
prot_seq = dna_seq.translate()

print("DNA sequence: ", dna_seq)
print("Protein sequence: ", prot_seq)
```

Q.1. Fetching sequences from NCBI and using BioPython

```python
#1

!pip install biopython
from Bio import Entrez, SeqIO

# Define the sequence ID (example: BRCA1 gene from human)
sequence_id = "NM_007294.4"

# Fetch the FASTA sequence from NCBI Nucleotide database
handle = Entrez.efetch(db="nucleotide", id=sequence_id, rettype="fasta", retmode="text")


# Read the FASTA record
record = SeqIO.read(handle, "fasta")

# Close the handle
handle.close()

# Print basic sequence information
print(record.id)
print(record.description)
print(len(record.seq))
print("\nFASTA format:\n")
print(record.format("fasta"))
```

```python
#Save to a local FASTA file
SeqIO.write(record, "downloaded_sequence.fasta", "fasta")
print("\nSequence saved as 'downloaded_sequence.fasta'")
```

Q.2. Counting Nucleotides in a DNA Sequence

```python
#2

# Define the DNA sequence
sequence = (
    "GCTGAGACTTCCTGGACGGGGGACAGGCTGTGGGGTTTCTCAGATAACTGGGCCCCTGCG"
    "CTCAGGAGGCCTTCACCCTCTGCTCTGGGTAAAGTTCATTGGAACAGAAAGAAATGGATT"
    "TATCTGCTCTTCGCGTTGAAGAAGTACAAAATGTCATTAATGCTATGCAGAAAATCTTAG"
    "AGTGTCCCATCTGTCTGGAGTTGATCAAGGAACCTGTCTCCACAAAGTGTGACCACATAT"
    "TTTGCAAATTTTGCATGCTGAAACTTCTCAACCAGAAGAAAGGGCCTTCACAGTGTCCTT"
    "TATGTAAGAATGATATAACCAAAAGGAGCCTACAAGAAAGTACGAGATTTAGTCAACTTG"
    "TTGAAGAGCTATTGAAAATCATTTGTGCTTTTCAGCTTGACACAGGTTTGGAGTATGCAA"
    "ACAGCTATAATTTTGCAAAAAAGGAAAATAACTCTCCTGAACATCTAAAAGATGAAGTTT"
    "CTATCATCCAAAGTATGGGCTACAGAAACCGTGCCAAAAGACTTCTACAGAGTGAACCCG"
    "AAAATCCTTCCTTGCAGGAAACCAGTCTCAGTGTCCAACTCTCTAACCTTGGAACTGTGA"
)

# Define the function to count nucleotides using a dictionary
def count_nucleotides(seq):
    counts = {
        'A': seq.count('A'),
        'T': seq.count('T'),
        'C': seq.count('C'),
        'G': seq.count('G')
    }
    return counts

# Call the function and print the result
nucleotide_counts = count_nucleotides(sequence)
print("Nucleotide Counts:")
for base, count in nucleotide_counts.items():
    print(f"{base}: {count}")
```

Q.3. Creating a Genus Abundance Heatmap Across Samples

```python
#3

import pandas as pd

# Create a dictionary with tuples with abundance data
data = {
    "Genus": ["Bacillus", "Pseudomonas", "Streptococcus", "Escherichia", "Clostridium"],
    "Sample_A": [50, 20, 15, 10, 5],
    "Sample_B": [40, 30, 25, 5, 10],
    "Sample_C": [60, 10, 35, 15, 20]
}

# Convert to DataFrame and set Genus as index
df = pd.DataFrame(data).set_index("Genus")
print(df)
print("\n")

import seaborn as sns
import matplotlib.pyplot as plt

# Create the plot using sns and plt
#uses normal scale
plt.figure(figsize=(7,5))
sns.heatmap(df, cmap="viridis", linewidths=0.6, cbar_kws={'label': 'Abundance'})
plt.title("Genus Abundance Heatmap Across Samples", fontsize=14)
plt.xlabel("Samples")
plt.ylabel("Genus")
plt.show()
```

Q. 4. Clustered Heatmap

```python
#4

import pandas as pd

# Create a dictionary with tuples with abundance data
data = {
    "Genus": ["Bacillus", "Pseudomonas", "Streptococcus", "Escherichia", "Clostridium"],
    "Sample_A": [50, 20, 15, 10, 5],
    "Sample_B": [40, 30, 25, 5, 10],
    "Sample_C": [60, 10, 35, 15, 20]
}

# Convert to DataFrame and set Genus as index
df = pd.DataFrame(data).set_index("Genus")
print(df)
print("\n")

# Use seaborn clustermap - generates a heatmap with dendogram
sns.clustermap(
    df,
    cmap="magma",             #colour palette
    linewidth=0.6,            #linnes between cells
    standard_scale=1,         #normalalize data across columns
    figsize=(8,6),            #figure size
    cbar_kws={'label' : 'Abundance'}          #label
)

plt.suptitle("Clustered Heatmap of Genus Abundance", fontsize=14, y=1.02)       #suptitle = super title - same as title
plt.show()
```

Q.5. Mean Abundance Barplot

```python
#5

df['Mean Abundance'] = df.mean(axis = 1)

plt.figure(figsize = (7,5))
sns.barplot(x = df.index, y = 'Mean Abundance', data = df, palette = 'coolwarm')
plt.title("Mean Genus Abundance", fontsize=14)
plt.xlabel("Genus")
plt.ylabel("Mean Abundance")
plt.ylim(0, 70)
plt.show()
```

Q.6. Calculate mean and Standard deviation (Error Barplot)

```python
#6

means = df[['Sample_A','Sample_B','Sample_C']].mean(axis=1)
stds = df[['Sample_A','Sample_B','Sample_C']].std(axis=1)
#.std is a python function to calculate standard deviation, axis =1 is for column values

plt.figure(figsize = (7,5))
plt.bar(df.index, means, yerr = stds, capsize = 5, color = 'skyblue', edgecolor = 'black')
plt.title("Genus Abundance with Error Bars", fontsize=14)
plt.xlabel("Genus")
plt.ylabel("Abundance(Mean +- Std)")
plt.ylim(0, 70)
plt.show()
```
